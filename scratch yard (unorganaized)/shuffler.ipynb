```python
import pandas as pd

# Load the data from the existing CSV file
data = pd.read_csv('4000checking2.csv')

# Shuffle the rows of the DataFrame
shuffled_data = data.sample(frac=1).reset_index(drop=True)

# Save the shuffled data to a new CSV file
shuffled_data.to_csv('checking2.csv', index=False)

print("Data has been shuffled and saved to 'checking2.csv'.")
```
